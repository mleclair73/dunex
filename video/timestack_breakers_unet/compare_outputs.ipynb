```python
import numpy as np
import matplotlib.pyplot as plt
from glob import glob
import os
import sys
sys.path.append('..')  # Add parent directory to Python path
from notebooks.video_io import get_datetime, load_frames

# Define all dates to process
DATES = ['20241118', '20241119', '20241120', '20241128', '20241203']

# Create necessary directories
for date in DATES:
    os.makedirs(f'results/diff/{date}', exist_ok=True)
os.makedirs('results/combined', exist_ok=True)

def process_folder(folder, fps=2):
    """
    Process a folder and return a dictionary of processed data.
    Data is normalized by length of breakers and FPS.
    
    Args:
        folder (str): Folder name (YYYYMMDD format)
        fps (int): Frames per second for normalization
    """
    data_dict = {}
    for fname in glob(f'results/{folder}/*breakers*'):
        breakers = load_frames(fname)
        count = (breakers!=0).sum(0)
        # Normalize by number of frames and FPS to get events per second
        normalized_count = count / (len(breakers) / fps)
        datetime_key = get_datetime(fname)
        # Ensure consistent shape
        if normalized_count.shape[0] == 1600:
            normalized_count = normalized_count.T
        data_dict[datetime_key] = normalized_count
    return data_dict

def create_sequential_diff_plot(data_dicts, datetime_key):
    """
    Create a figure showing sequential differences between all datasets.
    
    Args:
        data_dicts (dict): Dictionary mapping dates to their respective data dictionaries
        datetime_key (str): The datetime key to plot
    """
    n_comparisons = len(DATES) - 1
    fig, axes = plt.subplots(n_comparisons, 1, figsize=(24, 8 * n_comparisons))
    fig.suptitle(f'Sequential Differences - {datetime_key}', fontsize=16)
    
    for idx in range(n_comparisons):
        date1, date2 = DATES[idx], DATES[idx + 1]
        
        if datetime_key not in data_dicts[date1] or datetime_key not in data_dicts[date2]:
            axes[idx].text(0.5, 0.5, 'Data not available', ha='center', va='center')
            continue
            
        diff = data_dicts[date2][datetime_key] - data_dicts[date1][datetime_key]
        
        # Calculate symmetric limits based on percentiles
        lower = np.percentile(diff, 0.01)
        upper = np.percentile(diff, 99.99)
        abs_max = max(abs(lower), abs(upper))
        vmin, vmax = -abs_max, abs_max
        
        im = axes[idx].imshow(diff, cmap='bwr', vmin=vmin, vmax=vmax)
        plt.colorbar(im, ax=axes[idx], label='Difference in breakers per second')
        axes[idx].set_title(f'{date2} - {date1}\nColor limits: {vmin:.3f} to {vmax:.3f}')
    
    plt.tight_layout()
    plt.savefig(f'results/diff/sequential_diff_{datetime_key}.png')
    plt.close()

def create_combined_raw_plot(data_dicts, datetime_key):
    """
    Create a figure showing all raw datasets.
    
    Args:
        data_dicts (dict): Dictionary mapping dates to their respective data dictionaries
        datetime_key (str): The datetime key to plot
    """
    fig, axes = plt.subplots(len(DATES), 1, figsize=(32, 8 * len(DATES)))
    fig.suptitle(f'Raw Data Comparison - {datetime_key}', fontsize=16)
    
    # Find global min and max for consistent colormap
    all_data = []
    for date in DATES:
        if datetime_key in data_dicts[date]:
            all_data.append(data_dicts[date][datetime_key])
    
    if all_data:
        vmin, vmax = 0, np.percentile(np.concatenate([d.flatten() for d in all_data]), 99.99)
    
        for ax, date in zip(axes, DATES):
            if datetime_key in data_dicts[date]:
                im = ax.imshow(data_dicts[date][datetime_key], cmap='viridis', vmin=vmin, vmax=vmax)
                plt.colorbar(im, ax=ax, label='breakers per second')
                ax.set_title(f'{date}')
            else:
                ax.text(0.5, 0.5, 'Data not available', ha='center', va='center')
    
    plt.tight_layout()
    plt.savefig(f'results/combined/raw_{datetime_key}.png')
    plt.close()

# Process all folders
data_dicts = {date: process_folder(date) for date in DATES}

# Create visualizations for each datetime
all_datetime_keys = set()
for data_dict in data_dicts.values():
    all_datetime_keys.update(data_dict.keys())

for datetime_key in sorted(all_datetime_keys):
    create_sequential_diff_plot(data_dicts, datetime_key)
    create_combined_raw_plot(data_dicts, datetime_key)
```
